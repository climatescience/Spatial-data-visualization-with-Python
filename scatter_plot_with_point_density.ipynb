```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde, linregress

# 1. Load Excel Data
excel_file = 'master_scalled_rect.xlsx'
df = pd.read_excel(excel_file)

# 2. Define configuration for the 4 panels
panels = [
    {
        'col_x': 'NR_era', 'col_y': 'NR_modis',
        'xlabel': 'ERA $R_N$', 'ylabel': 'MODIS $R_N$',
        'tag': '(a)'
    },
    {
        'col_x': 'LE_era', 'col_y': 'LE_modis',
        'xlabel': 'ERA LE', 'ylabel': 'MODIS LE',
        'tag': '(b)'
    },
    {
        'col_x': 'SH_era', 'col_y': 'SH_modis',
        'xlabel': 'ERA SH', 'ylabel': 'MODIS SH',
        'tag': '(c)'
    },
    {
        'col_x': 'SH_era', 'col_y': 'SH_Bulk',
        'xlabel': 'ERA SH', 'ylabel': 'Bulk SH',
        'tag': '(d)'
    }
]

# 3. Initialize 2x2 Subplots
fig, axes = plt.subplots(2, 2, figsize=(13, 11))
axes = axes.flatten()

for ax, p in zip(axes, panels):
    # Extract clean data without NaNs
    data = df[[p['col_x'], p['col_y']]].dropna()
    x = data[p['col_x']].values
    y = data[p['col_y']].values

    # Calculate Gaussian Point Density
    xy = np.vstack([x, y])
    z = gaussian_kde(xy)(xy)

    # Sort points so highest density points are plotted on top
    idx = z.argsort()
    x_s, y_s, z_s = x[idx], y[idx], z[idx]

    # Plot Density Scatter
    sc = ax.scatter(x_s, y_s, c=z_s, s=15, cmap='jet', edgecolors='none')

    # Linear Regression
    slope, intercept, r_val, p_val, std_err = linregress(x, y)
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = slope * x_line + intercept
    ax.plot(x_line, y_line, color='red', linewidth=1.8, label='Regression Fit')

    # Metrics
    bias = np.mean(y - x)
    rmse = np.sqrt(np.mean((y - x)**2))
    r2 = r_val**2

    # Formatting Annotation Box
    stats_text = (
        f"y = {intercept:.2f} + {slope:.2f}*x\n"
        f"Pearson's r: {r_val:.2f}\n"
        f"Bias: {bias:+.4f}\n"
        f"RMSE: {rmse:.4f}\n"
        f"$R^2$: {r2:.4f}"
    )

    ax.text(
        0.05, 0.95, stats_text,
        transform=ax.transAxes,
        verticalalignment='top',
        fontsize=10,
        bbox=dict(boxstyle='round,pad=0.4', facecolor='white', alpha=0.85, edgecolor='gray')
    )

    # Panel Tag
    ax.text(0.90, 0.08, p['tag'], transform=ax.transAxes, fontsize=15, fontweight='bold')

    # Labels & Grid
    ax.set_xlabel(p['xlabel'], fontsize=12)
    ax.set_ylabel(p['ylabel'], fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.5)

    # Add Individual Colorbar
    cbar = plt.colorbar(sc, ax=ax)
    cbar.set_label('Point density', fontsize=11)

plt.tight_layout()

# Save output high-res figure
plt.savefig('4_panel_density_correlation new.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde, linregress

# 1. Load Excel Data
excel_file = 'master_scalled_rect.xlsx'
df = pd.read_excel(excel_file)

# 2. Define panels setup
panels = [
    {'col_x': 'NR_era', 'col_y': 'NR_modis', 'xlabel': r'ERA $R_N$', 'ylabel': r'MODIS $R_N$', 'tag': '(a)'},
    {'col_x': 'LE_era', 'col_y': 'LE_modis', 'xlabel': 'ERA LE', 'ylabel': 'MODIS LE', 'tag': '(b)'},
    {'col_x': 'SH_era', 'col_y': 'SH_modis', 'xlabel': 'ERA SH', 'ylabel': 'MODIS SH', 'tag': '(c)'},
    {'col_x': 'SH_era.1', 'col_y': 'SH_Bulk', 'xlabel': 'ERA SH', 'ylabel': 'Bulk SH', 'tag': '(d)'}
]

# 3. Compute Density for all panels to find global min/max for a unified scale
all_densities = []
panel_data = []

for p in panels:
    clean_df = df[[p['col_x'], p['col_y']]].dropna()
    x = clean_df[p['col_x']].values
    y = clean_df[p['col_y']].values

    # 2D Kernel Density Estimation
    xy = np.vstack([x, y])
    z = gaussian_kde(xy)(xy)

    all_densities.extend(z)
    panel_data.append((x, y, z))

# Unified scale bounds
vmin, vmax = min(all_densities), max(all_densities)

# 4. Create Subplots
fig, axes = plt.subplots(2, 2, figsize=(13, 10))

for ax, p, (x, y, z) in zip(axes.flatten(), panels, panel_data):
    # Sort points so dense ones plot on top
    idx = z.argsort()
    x_s, y_s, z_s = x[idx], y[idx], z[idx]

    # Plot using common vmin and vmax across all panels
    sc = ax.scatter(x_s, y_s, c=z_s, s=16, cmap='jet', vmin=vmin, vmax=vmax, edgecolors='none')

    # Regression Line
    slope, intercept, r_val, p_val, std_err = linregress(x, y)
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = slope * x_line + intercept
    ax.plot(x_line, y_line, color='red', linewidth=1.8)

    # Stats Text
    bias = np.mean(y - x)
    rmse = np.sqrt(np.mean((y - x)**2))
    r2 = r_val**2

    stats_text = (
        f"y = {intercept:.2f} + {slope:.2f}*x\n"
        f"Pearson's r: {r_val:.2f}\n"
        f"Bias: {bias:+.4f}\n"
        f"RMSE: {rmse:.4f}\n"
        f"$R^2$: {r2:.4f}"
    )

    ax.text(
        0.05, 0.95, stats_text,
        transform=ax.transAxes,
        verticalalignment='top',
        fontsize=9.5,
        bbox=dict(boxstyle='round,pad=0.4', facecolor='white', alpha=0.85, edgecolor='gray')
    )

    ax.text(0.90, 0.08, p['tag'], transform=ax.transAxes, fontsize=14, fontweight='bold')
    ax.set_xlabel(p['xlabel'], fontsize=11)
    ax.set_ylabel(p['ylabel'], fontsize=11)
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

# 5. Adjust subplots to fit full-height colorbar on the left
fig.subplots_adjust(left=0.12)

# Create a single vertical axis on the far-left spanning full height
cbar_ax = fig.add_axes([0.03, 0.08, 0.02, 0.88])
cbar = fig.colorbar(sc, cax=cbar_ax)

# Position colorbar ticks & label on the left side
cbar_ax.yaxis.set_ticks_position('left')
cbar_ax.yaxis.set_label_position('left')
cbar.set_label('Point density', fontsize=12, fontweight='bold')

plt.savefig('4_panel_single_colorbar_left.png', dpi=300, bbox_inches='tight')
plt.show()
```
