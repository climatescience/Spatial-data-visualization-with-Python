```python
# Install rasterio if needed
!pip install rasterio -q

import rasterio
import glob
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch
import os


# Read first 5 TIFF images
files = sorted(glob.glob("/content/*.tif"))[:5]

print("Images found:")
for f in files:
    print(os.path.basename(f))


# -----------------------------------
# Find unique classes (excluding 0)
# -----------------------------------

all_classes = []

for file in files:
    with rasterio.open(file) as src:
        img = src.read(1)

        if src.nodata is not None:
            img = np.where(img == src.nodata, np.nan, img)

        classes = np.unique(img[~np.isnan(img)])
        classes = classes[classes != 0]   # remove background class

        all_classes.extend(classes)


all_classes = np.unique(all_classes)

print("Classes used:", all_classes)


# Create colors only for real classes
#colors = plt.cm.tab20(np.linspace(0, 1, len(all_classes)))
#cmap = ListedColormap(colors)

# -----------------------------------
# Manually assign colors to classes
# -----------------------------------

class_colors = {
    1: "#228B22",   # Green - Cropland
    2: "#D73027",   # Red - Built-up
    3: "#F4A460",   # Orange - Bare land
    4: "#8B4513",   # Brown - Soil/Sparse vegetation
    5: "#BDBDBD"    # Grey - Others
}

# Create colors in the same order as raster classes
colors = [class_colors[int(c)] for c in all_classes]

print("Classes:", all_classes)
print("Colors assigned:", colors)

cmap = ListedColormap(colors)


# Class-value to color index
class_dict = {value: i for i, value in enumerate(all_classes)}


# -----------------------------------
# Plot 3 rows x 2 columns
# -----------------------------------

fig, axes = plt.subplots(
    3, 2,
    figsize=(8, 10),
    constrained_layout=True
)

axes = axes.flatten()


for i, (ax, file) in enumerate(zip(axes, files)):

    with rasterio.open(file) as src:
        img = src.read(1)

        if src.nodata is not None:
            img = np.where(img == src.nodata, np.nan, img)


    # Create empty array
    classified = np.full(img.shape, np.nan)


    # Assign colors only to valid classes
    for value, idx in class_dict.items():
        classified[img == value] = idx


    # Plot
    labels = ['(a)', '(b)', '(c)', '(d)', '(e)']

    ax.imshow(
        classified,
        cmap=cmap,
        interpolation="nearest"
    )

    ax.set_title(
        labels[i],
    fontsize=14,
    loc='left',
    fontweight='bold'
    )

    ax.axis("off")


# Remove unused panel (6th position)
for ax in axes[len(files):]:
    ax.axis("off")


all_classes = np.unique(all_classes)
print("Actual raster classes:", all_classes)



# -----------------------------------
# Legend (without class 0)
# -----------------------------------
class_names = {
    1: "Cropland",
    2: "Built-up",
    3: "Bare land",
    4: "Water",
    5: "Others"
}


legend_elements = [
    Patch(
        facecolor=colors[i],
        label=class_names[int(v)]
    )
    for i, v in enumerate(all_classes)
]


fig.legend(
    handles=legend_elements,
    loc="lower center",
    ncol=5,
    fontsize=10,
    bbox_to_anchor=(0.5, -0.05)
)


plt.show()
```
